# Question 6 — Median horsepower before and after filling missing values

```python
median_before = df["horsepower"].median()
median_before # median horsepower before filling missing values
```

```python
mode_horsepower = df["horsepower"].mode()[0]
mode_horsepower # median horsepower after filling missing values with the mode
```

```python
df_filled = df.copy() # making a copy of the original dataframe to fill missing values without altering the original data
```

```python
df_filled["horsepower"] = df_filled["horsepower"].fillna(
    mode_horsepower
) # fill missing values in the "horsepower" column with the mode value
```

```python
median_after = df_filled["horsepower"].median()
median_after # median horsepower after filling missing values with the mode
```

```python
median_before, median_after # compare the median horsepower before and after filling missing values
```

# Question 7 

```python
# question 7 important
asia_subset = df[df["origin"] == "Asia"][
    ["vehicle_weight", "model_year"]
].head(7)

asia_subset # display the first 7 rows of the filtered dataframe containing only Asian cars and the specified columns "vehicle_weight" and "model_year"
```

```python
X = asia_subset.values
X # display the values of the filtered dataframe as a NumPy array
```

```python
X.shape # display the shape of the NumPy array (number of rows and columns) 
```

```python
XTX = X.T @ X
XTX # compute the dot product of the transpose of the NumPy array X and X itself, resulting in a new matrix XTX here @ is the matrix multiplication operator in Python, which is used to perform matrix multiplication between two arrays. In this case, X.T is the transpose of the array X, and @ is used to multiply it with X, resulting in a new matrix XTX.
```

```python
XTX_inv = np.linalg.inv(XTX)
XTX_inv # compute the inverse of the matrix XTX using NumPy's linear algebra module (np.linalg) and the inv() function. The result is stored in the variable XTX_inv, which represents the inverse of the matrix XTX.
```

```python
y = np.array(
    [1100, 1300, 800, 900, 1000, 1100, 1200]
) # create a NumPy array y containing the target values corresponding to the rows in the asia_subset dataframe
```

```python
w = XTX_inv @ X.T @ y
w # compute the weights (coefficients) for a linear regression model using the normal equation. The weights are calculated by multiplying the inverse of the dot product of X transpose and X (XTX_inv) with the dot product of X transpose and the target values y. The result is stored in the variable w, which represents the coefficients for the linear regression model.
```
